```python
#Q3
missed_pickups_by_hauler = (missed_pickups_df
    .groupby('Trash Hauler')['fine_owed'] 
    .sum() 
    .reset_index())
                            
missed_pickups_by_hauler.columns = ['Trash Hauler', 'total_fine_owed']

missed_pickups_by_hauler = missed_pickups_by_hauler.sort_values(
    by='total_fine_owed', ascending=False)

missed_pickups_by_hauler
```

```python
#Q4
missed_pickups_by_route = (missed_pickups
    .groupby(['Trash Route', 'Trash Hauler'])
    .size()
    .reset_index(name='total_missed'))

missed_pickups_by_route = missed_pickups_by_route.sort_values(
    by='total_missed', ascending=False)

missed_pickups_by_route
```
